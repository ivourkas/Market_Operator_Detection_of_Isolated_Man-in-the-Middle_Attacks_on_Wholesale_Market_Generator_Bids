```python
import numpy as np
import pandas as pd

# Load the dataset
df = pd.read_csv('data/N57_G7.csv', header=None)
```

```python
# Transpose the DataFrames
df = df.T


# Generate column names
feature_columns = [f"ft{i}" for i in range(1, 6481)]
remaining_columns = ["fval", 
                     "ssn", "type_of_attack", "gen_attacked"]
all_columns = feature_columns + remaining_columns

# Assign column names to the DataFrames
df.columns = all_columns

```

```python
# Export to new csv to check the column names
df.to_csv('data/N57_G7_transposed.csv', header=df.columns, index=False)
df = pd.read_csv('data/N57_G7_transposed.csv')
df.shape

# Count how many attack types per ssn
attack_counts = df.groupby('ssn')['type_of_attack'].value_counts().unstack(fill_value=0)
print(attack_counts)
```

```python
# Drop NaN values

# Filter rows where all features are zero
all_zero_rows = (df[feature_columns]==-100000).all(axis=1)
df = df[~all_zero_rows] # Keep rows where at least one feature is non-zero

df.shape
```
